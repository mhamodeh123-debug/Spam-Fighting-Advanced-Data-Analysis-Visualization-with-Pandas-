```python
import email
import os
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
def flatten_to_string(parts):
    ret = []
    if type(parts) == str:
        ret.append(parts)
    elif type(parts) == list:
        for part in parts:
            ret += flatten_to_string(part)
    elif parts.get_content_type == 'text/plain':
        ret += parts.get_payload()
    return ret
```

```python
def extract_email_text(path):
    # Load a single email from an input file
    with open(path, 'r', errors='ignore') as f:
        # read the email and store it in msg
        msg = email.message_from_file(f)
    if not msg: return ""
    # Read the email subject
    subject = msg['Subject']
    if not subject: subject = ""
    # Read the email body
    body = ' '.join(m for m in flatten_to_string(msg.get_payload()) if type(m) == str)
    if not body: body = ""
    return subject + ' ' + body
```

```python
def read_email_files():
    X = []
    y = []
    for i in range(len(labels)):
        filename = 'inmail.' + str(i+1)
        # Extract subject and body text from a single email file
        email_str = extract_email_text(os.path.join(DATA_DIR, filename))
        X.append(email_str)
        y.append(labels[filename])
    return X, y
```

```python
def classify_email(email, model, vectorizer):
    email_vectorized = vectorizer.transform([email])
    prediction = model.predict(email_vectorized)
    return prediction[0]
```

```python
DATA_DIR = '../part1/trec07p/data'
LABELS_FILE = '../part1/trec07p/full/index'
labels = {}
# Read the labels
with open(LABELS_FILE) as f:
    for line in f:
        line = line.strip()
        label, key = line.split()
        labels[key.split('/')[-1]] = 1 if label.lower() == 'ham' else 0
X, y = read_email_files()
# X: list of string, each is the subject + body of the corresponding email
# y: list of all labels (0:spam, 1:ham)
```

```python
#1 (80:20) train:test ratio
X_train1, X_test1, y_train1, y_test1= train_test_split(X, y, train_size=0.8)
#2 (60:40) train:test ratio
X_train2, X_test2, y_train2, y_test2= train_test_split(X, y, train_size=0.6)
#3 (50:50) train:test ratio
X_train3, X_test3, y_train3, y_test3= train_test_split(X, y, train_size=0.5)
#4 (40:60) train:test ratio
X_train4, X_test4, y_train4, y_test4= train_test_split(X, y, train_size=0.4)
#5 (20:80) train:test ratio
X_train5, X_test5, y_train5, y_test5= train_test_split(X, y, train_size=0.2)
```

```python
vectorizer1 = CountVectorizer()
vectorizer2 = CountVectorizer()
vectorizer3 = CountVectorizer()
vectorizer4 = CountVectorizer()
vectorizer5 = CountVectorizer()
X_train_vector1 = vectorizer1.fit_transform(X_train1)
X_train_vector2 = vectorizer2.fit_transform(X_train2)
X_train_vector3 = vectorizer3.fit_transform(X_train3)
X_train_vector4 = vectorizer4.fit_transform(X_train4)
X_train_vector5 = vectorizer5.fit_transform(X_train5)
```

```python
X_test_vector1 = vectorizer1.transform(X_test1)
X_test_vector2 = vectorizer2.transform(X_test2)
X_test_vector3 = vectorizer3.transform(X_test3)
X_test_vector4 = vectorizer4.transform(X_test4)
X_test_vector5 = vectorizer5.transform(X_test5)
```

```python
# Initialize the classifier and make label predictions
mnb1 = MultinomialNB()
mnb2 = MultinomialNB()
mnb3 = MultinomialNB()
mnb4 = MultinomialNB()
mnb5 = MultinomialNB()
mnb1.fit(X_train_vector1, y_train1)
mnb2.fit(X_train_vector2, y_train2)
mnb3.fit(X_train_vector3, y_train3)
mnb4.fit(X_train_vector4, y_train4)
mnb5.fit(X_train_vector5, y_train5)
y_pred1 = mnb1.predict(X_test_vector1)
y_pred2 = mnb2.predict(X_test_vector2)
y_pred3 = mnb3.predict(X_test_vector3)
y_pred4 = mnb4.predict(X_test_vector4)
y_pred5 = mnb5.predict(X_test_vector5)
# Print results
a1='Accuracy {:.1%}'.format(accuracy_score(y_test1, y_pred1))
a2='Accuracy {:.1%}'.format(accuracy_score(y_test2, y_pred2))
a3='Accuracy {:.1%}'.format(accuracy_score(y_test3, y_pred3))
a4='Accuracy {:.1%}'.format(accuracy_score(y_test4, y_pred4))
a5='Accuracy {:.1%}'.format(accuracy_score(y_test5, y_pred5))
print(a1,a2,a3,a4,a5,sep="\n")
```
